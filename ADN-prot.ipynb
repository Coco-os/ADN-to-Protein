```python
from Bio.Seq import Seq

secuencia = Seq("ATGGCGTTATTGCATGTAATTAACCGTTAGCT")

complementaria = secuencia.complement()

print("Original:       ", secuencia)
print("Complementaria: ", complementaria)
```

```python
from Bio import SeqIO

registro = SeqIO.read("pseudomonas.fasta", "fasta")

adn = registro.seq

print("ADN:  ", adn)
print("ARNm: ", adn.transcribe())
```
