```python
import requests
from Bio.Seq import Seq

# Gen FGFR2 en humano
gene_id = "ENSG00000066468"

url = f"https://rest.ensembl.org/lookup/id/{gene_id}"

params = {
    "expand": 1,
    "content-type": "application/json"
}

response = requests.get(url, params=params)
response.raise_for_status()

gene = response.json()

print("Gen:", gene["display_name"])
print("ID:", gene["id"])
print("Transcrito canónico:", gene.get("canonical_transcript"))

print("\nTranscritos:")

for transcript in gene.get("Transcript", []):
    print(
        transcript["id"],
        transcript.get("biotype"),
        transcript.get("Translation", {}).get("id")
    )
```
